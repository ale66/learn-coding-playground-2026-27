#### Step 4 · download messages

Run once with the example channel

Download the *latest* N= 200 messages. 

Long-term channels like these have lots of posts.
E.G., 'ddgeopolitics' has about 200,000 posts: don't use `None` (= all).

````python
N = 200
````

````python
rows = []

# a structure similar to reading files 
# as we are connecting via internet, possible delays are accommodated by the 'async' command 
# rows are appended as they arrive
async for m in client.iter_messages(CHANNEL, limit=N):

    rows.append({
        "date": m.date,
        "text": m.text,
        "views": m.views,
        "link": f"https://t.me/{CHANNEL}/{m.id}",
    })
````

````python
# put the rows into a Pandas dataframe
df = pd.DataFrame(rows)

df.head()
````

What are we getting from the Telegram API?

| Column | What it contains |
|--------|------------------|
| `date` | day and time of the post (UTC) |
| `text` | the text of the message |
| `views` | how many times it was seen |
| `link` | opens the message on Telegram |
